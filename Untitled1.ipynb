#Library

```python
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import cifar100
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.utils import to_categorical, plot_model
```

#Loading the Dataset

```python
# Load dataset
(x_train, y_train), (x_test, y_test) = cifar100.load_data()
```

```python
# Print dataset shapes
print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)
```

# Preprocess CIFAR-100

```python
# Normalize pixel values
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# One-hot encode labels
y_train = to_categorical(y_train, 100)
y_test = to_categorical(y_test, 100)

print(x_train.shape, y_train.shape)
```

#Custom CNN (Part 1)

```python
custom_cnn = keras.Sequential([
    layers.Conv2D(32, (3,3), padding='same', activation='relu',
                   kernel_initializer='he_normal', input_shape=(32,32,3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2,2)),
    layers.Dropout(0.25),

    layers.Conv2D(64, (3,3), padding='same', activation='relu',
                   kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2,2)),
    layers.Dropout(0.25),

    layers.Conv2D(128, (3,3), padding='same', activation='relu',
                   kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2,2)),
    layers.Dropout(0.25),

    layers.Flatten(),
    layers.Dense(256, activation='relu', kernel_initializer='he_normal'),
    layers.Dropout(0.5),
    layers.Dense(100, activation='softmax')
])

custom_cnn.summary()
```

```python
#combiling model
custom_cnn.compile(optimizer='adam',
                    loss='categorical_crossentropy',
                    metrics=['accuracy'])

# Train the model
history_custom = custom_cnn.fit(
    x_train, y_train,
    validation_split=0.2,
    epochs=5,
    batch_size=64,
)
```

```python
# Evaluate the model
custom_loss, custom_acc = custom_cnn.evaluate(x_test, y_test)

# Print results
print("Custom CNN Test Accuracy:", custom_acc)
```

#VGG-16 (Part 2)

```python
vgg16_base = keras.applications.VGG16(
    weights=None,
    include_top=False,
    input_shape=(32,32,3)
)

vgg16 = keras.Sequential([
    vgg16_base,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(100, activation='softmax')
])

vgg16.summary()
```

```python
vgg16.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

history_vgg = vgg16.fit(
    x_train, y_train,
    validation_split=0.1,
    epochs=3,
    batch_size=64,)
```

```python
vgg_loss, vgg_acc = vgg16.evaluate(x_test, y_test)
print("VGG-16 Test Accuracy:", vgg_acc)
```

#Plot training curves

```python
def plot_curves(history, title):
    plt.plot(history.history['accuracy'], label='Train Acc')
    plt.plot(history.history['val_accuracy'], label='Val Acc')
    plt.title(title + ' - Accuracy')
    plt.xlabel('Epoch'); plt.legend(); plt.show()

    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Val Loss')
    plt.title(title + ' - Loss')
    plt.xlabel('Epoch'); plt.legend(); plt.show()

plot_curves(history_custom, "Custom CNN")
plot_curves(history_vgg, "VGG-16")
```

#Compare results

```python
results = pd.DataFrame({
    'Model': ['Custom CNN', 'VGG-16'],
    'Test Accuracy': [custom_acc, vgg_acc]
})
print(results)
```
